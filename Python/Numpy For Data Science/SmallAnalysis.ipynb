```python
import numpy as np
students = np.random.randint(0, 101, 500)
```

```python
print(students)
```

```python
print("Mean :",np.mean(students))
print("Median :",np.median(students))
print("Standard Deviation :",np.std(students))
print("Pass Percentage :",(np.sum(students>=40)/students.size)*100)
```

```python
A = np.sum(students >= 90)
B = np.sum((students>= 80) & (students < 90))
C = np.sum((students >= 70) & (students < 80))
D = np.sum((students >= 40) & (students < 70))
F = np.sum(students < 40)

print("A Grade:", A)
print("B Grade:", B)
print("C Grade:", C)
print("D Grade:", D)
print("F Grade:", F)
```

```python
top_10 = np.sort(students)
print(top_10[::-1][0:11])
```

```python
counts,bins = np.histogram(students,[0,21,41,61,81,200])
print("Counts :",counts)
print("Bins :",bins)
```
