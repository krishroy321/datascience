Built-In Mathematical Function in Numpy

```python
import numpy as np
```

```python
arr = np.array([3,5,2,6,789,100])
arr1 = np.array([1,2,3,4,5,6])
```

```python
np.mean(arr)
```

```python
np.std(arr)
```

```python
np.var(arr)
```

```python
np.min(arr)
```

```python
np.max(arr)
```

```python
np.sum(arr)
```

```python
np.prod(arr)
```

```python
np.median(arr)
```

```python
np.percentile(arr,50)
```

```python
np.argmin(arr)
```

```python
np.argmax(arr)
```

```python
np.corrcoef(arr,arr1)
```

```python
np.unique(arr)
```

```python
np.diff(arr)
```

```python
np.cumsum(arr)
```

```python
np.linspace(0,10,5)
```

```python
np.log(arr)
```

```python
np.exp(arr)
```
