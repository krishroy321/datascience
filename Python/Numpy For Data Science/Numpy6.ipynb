Broadcasting in Numpy

```python
import numpy as np
arr = np.array([1,2,3,4,5,6])
result = arr**2
print(result)
```

```python
result + 10
print(result)
```

```python
arr1 = np.array([1,2,3])
arr2 = np.array([10,20,30])

result1 = arr1+arr2
print(result1)
```

```python
arr1 = np.array([[1,2,3],[4,5,6]])
arr2 = np.array([1,2,3])

result1 = arr1 + arr2
print(result1)
```

Hands - on : Applying Broadcasting to Real-World Scenarios

```python
data = np.array([[10 , 20 ,30],
                 [15,25,35],
                 [20,30,40],
                 [25,35,45],
                 [30,40,50]])

mean = data.mean(axis=0)
print("Mean :",mean)
SD = data.std(axis=0)
print("Standard Deviation :",SD)
print("  ")
normalized_data = (data-mean)/SD
print(normalized_data)


```
