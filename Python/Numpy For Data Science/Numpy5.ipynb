```python
import numpy as np
arr = np.array([1,2,3,4,5])
arr.dtype
```

Changing Data Types

```python
arr1 = np.array([1.5,2.7,3.9])
print(arr1.dtype)

arr1_int = arr1.astype("int64")
print(arr1_int)
print(arr1_int.dtype)
```

DownCasting to Save Memory

```python
arr_large = np.array([1000000,2000000,3000000],dtype = "int64")
print(arr_large.dtype)
arr_small = arr_large.astype("int32")
print(arr_small.dtype)
```

```python
arr_int64 = np.array([1,2,3],dtype="int64")
arr_int32 = np.array([1,2,3],dtype="int32")

print(arr_int64.nbytes)
print(arr_int32.nbytes)
```

String Data Type in Numpy

```python
arr_str = np.array(['Apple','Banana','Cherry'],dtype='U10')
arr_str.dtype
```

Complex Numbers


```python
arr_cmplx = np.array([1 + 2j, 3 + 4j, 5 + 6j],dtype='complex64')
arr_cmplx.dtype
```

Object Data Type

```python
arr_obj  = np.array([{'a':1},[1,2,3],'hello'],dtype=object)
arr_obj.dtype
```
