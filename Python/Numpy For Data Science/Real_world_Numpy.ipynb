```python
import numpy as np

marks = np.random.randint(0, 101, 100)
```

```python
print(marks)
```

```python
print("Average Marks :",np.average(marks))
print("Highest Marks :",np.max(marks))
print("Lowest Marks :",np.min(marks))
print("No. Of Students Scoring Above 90 :",(np.sum(marks>90)))
print("No. Of Students Failing (<40) :",(np.sum(marks<40)))
print("No. Of Students Passing (>40) :",(np.sum(marks>40)))
print("Top 5 Marks :",(np.sort(marks)[::-1][0:6]))
print("Percentage of Student Passing :", (np.sum(marks >=40) / marks.size) * 100)
```
